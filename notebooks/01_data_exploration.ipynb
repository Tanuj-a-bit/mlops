# Phase 1: Data Understanding and Exploration

This notebook explores the RetailRocket dataset, providing statistical details and insights as required by the project specifications.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Set project root
import sys
sys.path.append('..')

from src.data.loader import DataLoader
```

## 1. Load Data
Acknowledge Source: RetailRocket Recommender System Dataset (Kaggle).

```python
loader = DataLoader(data_dir="../data/raw")
events = loader.load_events()
print(f"Total Interaction Events: {len(events):,}")
```

## 2. Event Distribution
Explore the frequency of views, add-to-carts, and transactions.

```python
plt.figure(figsize=(10, 6))
sns.countplot(data=events, x='event')
plt.title("Distribution of Event Types")
plt.show()
```

## 3. User & Item Statistics

```python
print(f"Unique Visitors: {events['visitorid'].nunique():,}")
print(f"Unique Items: {events['itemid'].nunique():,}")

items_per_user = events.groupby('visitorid')['itemid'].count()
print(f"Average interactions per user: {items_per_user.mean():.2f}")
```
